```python
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns
```

```python
IT_Department_A = {
    "Employee ID": ["E101", "E102", "E103", "E104", "E105", "E106"],
    "Employee Name": ["Rohan", "Kiran", "Aditya", "Sneha", "Pooja", "Rahul"],
    "Project": [75, 65, 73, 68, 72, 67]
}
IT_Department_B= {
    "Employee ID": ["E201", "E202", "E203", "E204", "E205", "E206"],
    "Employee Name": ["Arjun", "Deepak", "Kavya", "Megha", "Sanjay", "Varun"],
    "Project": [90, 47, 93, 47, 96, 51]
}

deptA_df = pd.DataFrame(IT_Department_A)
deptB_df = pd.DataFrame(IT_Department_B)
```

```python
range_deptA =np.max(deptA_df['Project']) - np.min(deptA_df['Project'])
range_deptB = np.max(deptB_df['Project']) - np.min(deptB_df['Project'])
```

```python
plt.scatter([i for i in range(0, len(deptA_df))], 
            deptA_df["Project"], color='blue', label='Department A',s=100,marker='o')
plt.scatter([i for i in range(0, len(deptB_df))], 
            deptB_df["Project"], color='red', label='Department B',s=100,marker='s')
plt.axhline(y=range_deptA.mean(), color='blue', linestyle='--', label='Range of Department A')
plt.axhline(y=range_deptB, color='red', linestyle='--', label='Range of Department B')
plt.legend()
plt.ylim(20,110)
plt.show()
```

```python
plt.scatter([i for i in range(0,len(deptA_df['Project']))], deptA_df['Project'], color='blue', label='Department A',s=100,marker='o')
plt.scatter([i for i in range(0,len(deptB_df['Project']))], deptB_df['Project'], color='red', label='Department B',s=100,marker='s')
plt.xlabel('Employee ID')
plt.ylabel('Project Score')
plt.title('Project Scores by Department')
plt.legend()
plt.ylim(20,110)
plt.axhline(y=deptA_df['Project'].mean(), color='blue', linestyle='--', label='Dept A Mean')
plt.axhline(y=deptB_df['Project'].mean(), color='red', linestyle='--', label='Dept B Mean')
plt.show()
```

```python
x_i = deptA_df['Project']
x_bar = deptA_df['Project'].mean()
n = len(deptA_df['Project'])
MAD_depA = np.sum(np.abs(x_i - x_bar)) / n
MAD_depA
```

```python
x_i = deptB_df['Project']
x_bar = deptB_df['Project'].mean()
n = len(deptB_df['Project'])
MAD_depB = np.sum(np.abs(x_i - x_bar)) / n
MAD_depB
```
