```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
```

```python
titanic = pd.read_csv("titanic.csv")
titanic.head(10)
```

```python
meanValue =np.mean(titanic['Fare'])
meanValue
```

```python
sns.histplot(data=titanic, x='Fare',bins=[i for i in range(0, 151, 10)])
plt.plot([meanValue for i in range(0, 351, 10)], [i for i in range(0, 351, 10)], c='red')
plt.ylim(0,400)
plt.show
```

```python
medianValue = np.median(titanic['Fare'])
medianValue
```

```python
sns.histplot(data=titanic, x='Fare',bins=[i for i in range(0, 151, 10)])
plt.plot([medianValue for i in range(0, 351, 10)], [i for i in range(0, 351, 10)], c='green')
plt.ylim(0,400)
plt.show
```

```python
modeValue = titanic['Fare'].mode()[0]
modeValue
```

```python
sns.histplot(data=titanic, x='Fare',bins=[i for i in range(0, 151, 10)])
plt.plot([modeValue for i in range(0, 351, 10)], [i for i in range(0, 351, 10)], c='blue')
plt.ylim(0,400)
plt.show
```

```python
sns.histplot(x='Fare', data=titanic, bins = [i for i in range(0, 151, 10)]
             )
plt.plot([meanValue for i in range(0 ,351)],[i for i in range(0 ,351)], color='red' , label = 'Mean')
plt.plot([medianValue for i in range(0 ,351)],[i for i in range(0 ,351)], color='black' , label = 'Median')
plt.plot([modeValue for i in range(0 ,351)],[i for i in range(0 ,351)], color='yellow' , label = 'Mode')
plt.legend()
plt.title('Measures of Central Tendency titanic fare' , fontsize = 15 , fontweight = 'bold')
plt.ylim(0, 400)
plt.tight_layout()
plt.show()
```
