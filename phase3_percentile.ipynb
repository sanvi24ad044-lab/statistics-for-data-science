```python

import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns
```

```python
titanic_df = pd.read_csv('titanic.csv')
titanic_df.head()
```

```python
titanic_df.isnull().sum()
```

```python
mean_age = titanic_df['Age'].mean()
titanic_df['Age'] = titanic_df['Age'].fillna(mean_age)
```

```python
print("0th percentile of Age:", np.percentile(titanic_df['Age'], 0))
print("25th percentile of Age:", np.percentile(titanic_df['Age'], 25))
print("50th percentile of Age:", np.percentile(titanic_df['Age'], 50))
print("75th percentile of Age:", np.percentile(titanic_df['Age'], 75))      
print("100th percentile of Age:", np.percentile(titanic_df['Age'], 100))    
```

```python
sns.boxplot(x='Age', data=titanic_df)
plt.show()
```

```python
mean_fare = titanic_df['Fare'].mean()
titanic_df['Fare'] = titanic_df['Fare'].fillna(mean_fare)


print("0%:", np.percentile(titanic_df['Age'], 0))
print("25%:", np.percentile(titanic_df['Age'], 25))
print("50%:", np.percentile(titanic_df['Age'], 50))
print("75%:", np.percentile(titanic_df['Age'], 75))
print("100%:", np.percentile(titanic_df['Age'], 100))   


sns.boxplot(x='Fare',data=titanic_df)
plt.show()
```
